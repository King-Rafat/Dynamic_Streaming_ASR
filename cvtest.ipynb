```python
import pandas as pd

df = pd.read_csv("test.tsv", sep = '\t')
```

```python
df
```

```python
df = df[df['up_votes'] >= 2]
```

```python
audio_path = df['path'].tolist()
sentence = df['sentence'].tolist()
```

```python
!mkdir test_cv
```

```python
import shutil
import os

source_dir = "C:\\MenoChat\\code\\code\\Menochat-Gemma\\Meno-Chat ASR\\mcv-scripted-bn-v23.0\\cv-corpus-23.0-2025-09-05\\bn\\clips\\"
destination_directory = 'test_cv\\'

os.makedirs(destination_directory, exist_ok=True)
for i in range(len(audio_path)):
    try:
        # Copy the file to the destination directory
        shutil.copy2(source_dir + audio_path[i], destination_directory)
        print(f"File '{source_dir}' copied successfully to '{destination_directory}'.")
    except shutil.SameFileError:
        print("Source and destination represent the same file.")
    except PermissionError:
        print("Permission denied.")
    except Exception as e:
        print(f"Error occurred: {e}")
```

```python
with open("test_cv_moz.scp", "w") as file:
    for i in range(len(audio_path)):
        file.write(audio_path[i] +" "+ "/test_cv/"+ audio_path[i]+"\n")


with open("test_text.txt", "w", encoding="utf-8") as file:
    for i in range(len(audio_path)):
        file.write(audio_path[i] +" "+ sentence[i]+"\n")
```
